# Python Basics

1. What is Python, and why is it popular?
   -> Python is a high-level, interpreted programming language known for its simplicity and readability. Created by Guido van Rossum and first released in 1991, Python emphasizes code readability through its clean syntax and indentation-based structure.
   It is popular because:
   Easy to learn with clean syntax.
   Versatile for web development, data science, AI, automation, etc.
   Rich libraries and frameworks.
   Cross-platform support.
   Large, active community.

2. What is an interpreter in Python?
   -> An interpreter in Python is a program that executes Python code line by line, translating it into machine-readable instructions at runtime. It handles lexical analysis, parsing, bytecode compilation, and execution. Python’s default interpreter is CPython, with alternatives like PyPy and Jython.

3. What are pre-defined keywords in Python?
   -> Pre-defined keywords in Python are reserved words with special meanings and cannot be used as identifiers (e.g., variable or function names). They define the syntax and structure of Python code.

   Examples of Keywords:
   Control Flow: if, else, elif, for, while, break, continue
   Data Types: True, False, None
   Functions: def, return, lambda
   Class/Objects: class, self, init
   Error Handling: try, except, finally, raise
   Others: import, with, as, global, nonlocal

4. Can keywords be used as variable names?
   -> No, keywords cannot be used as variable names in Python. They are reserved for specific purposes in the language's syntax and using them as identifiers will result in a SyntaxError.

5. What is mutability in Python?
   -> Mutability in Python refers to whether an object's state (i.e., its contents) can be changed after it is created.

   Mutable Objects:

   Their contents can be changed after creation.
   Examples: list, dict, set

6. Why are lists mutable, but tuples are immutable?
   -> The difference between lists and tuples in Python comes down to their design purpose and implementation:

   Why Lists Are Mutable:
   Purpose: Lists are designed for dynamic collections of items where frequent additions, deletions, or updates are needed.
   Implementation: Lists have a flexible memory structure that allows elements to be modified or resized without creating a new object.

   Why Tuples Are Immutable:
   Purpose: Tuples are meant for fixed, ordered collections of items where immutability ensures data integrity and safety.
   Implementation: Tuples are stored more compactly in memory, and immutability allows Python to optimize their performance (e.g., caching and hashability for use as dictionary keys or in sets).

7. What is the difference between “==” and “is” operators in Python?
   -> The == and is operators in Python are used for comparison, but they serve different purposes:

  1. == (Equality Operator):
  Compares the values of two objects to check if they are equal.
  Even if the objects are stored in different memory locations, == will return True if their values are the same.

  2. is (Identity Operator):
  Compares the identity of two objects (i.e., whether they refer to the same memory location).
  Returns True only if both objects are the same instance.

  Key Difference:
  == checks value equality.
  is checks reference equality (object identity).

8. What are logical operators in Python?
   -> Logical operators in Python are used to perform logical operations on Boolean values (True or False). They are essential for making decisions and controlling the flow of programs.

   1. and (Logical AND):
   Returns True if both operands are True, otherwise returns False.
   2. or (Logical OR):
   Returns True if at least one operand is True, otherwise returns False.
   3. not (Logical NOT):
   Reverses the Boolean value of the operand: returns True if the operand is False, and vice versa.

9. What is type casting in Python?
   -> Type casting in Python is the process of converting one data type into another.
   Common Type Casting Functions:
   int(): Converts to an integer.
   float(): Converts to a float.
   str(): Converts to a string.
   list(): Converts to a list.
   tuple(): Converts to a tuple.

10. What is the difference between implicit and explicit type casting?
   -> The difference between implicit and explicit type casting in Python lies in how the conversion of data types is handled:

   Key Difference:
   Implicit: Happens automatically and is safe, like when a smaller numeric type is promoted to a larger one.
   Explicit: Requires the programmer to manually call a conversion function, which can be prone to errors if done incorrectly.

11. What is the purpose of conditional statements in Python?
   -> Conditional statements in Python are used to control the flow of the program based on certain conditions. They allow the program to execute different blocks of code depending on whether a condition is True or False.

   Purpose of Conditional Statements:
   Decision Making:
   They enable the program to make decisions and choose which actions to take based on certain conditions.
   
   Control Flow:
   Conditional statements help direct the flow of execution by evaluating expressions and deciding which code block should run.

12. How does the elif statement work?
   -> The elif (short for "else if") statement in Python is used to check multiple conditions in a series.

   How elif Works:
   The program first checks the condition in the if statement.
   If the if condition is True, its code block runs, and the rest of the elif or else blocks are skipped.
   If the if condition is False, the program moves to the first elif condition.
   If any elif condition is True, its code block runs, and the rest of the conditions are ignored.
   If none of the if or elif conditions are True, the program will execute the else block (if present).

13. What is the difference between for and while loops?
  -> The difference between for and while loops in Python lies in how they are used to iterate or repeat a block of code.

   Key Differences:
   For Loop: Used when you know the number of iterations or when iterating over a sequence.
   While Loop: Used when the number of iterations is not fixed and depends on a condition.

14. Describe a scenario where a while loop is more suitable than a for
    loop.
   -> A while loop is more suitable than a for loop when you don't know beforehand how many times the loop needs to execute, and you want the loop to continue until a certain condition is met.

   Scenario Example: Waiting for User Input
   Imagine you are building a program that prompts a user for their password and keeps asking until they enter the correct one.

   Why a while loop is suitable:
   You don't know how many attempts the user will make before they enter the correct password, so the loop needs to run indefinitely until a valid password is entered.
   The condition for exiting the loop depends on the user’s input.

```python
# Ans. 6

def demonstrate_mutable_immutable():
    # Mutable argument (list)
    mutable_list = [1, 2, 3]

    # Immutable argument (integer)
    immutable_num = 10

    # Function to modify the arguments
    def modify_arguments(mutable_arg, immutable_arg):
        # Modify the mutable argument (list)
        mutable_arg.append(4)

        # Try to modify the immutable argument (integer)
        immutable_arg += 5

        print("Inside the function:")
        print("Modified mutable argument (list):", mutable_arg)
        print("Modified immutable argument (integer):", immutable_arg)

    # Call the function with the mutable and immutable arguments
    modify_arguments(mutable_list, immutable_num)

    # Show the state after the function call
    print("\nAfter the function call:")
    print("Mutable argument (list) outside function:", mutable_list)
    print("Immutable argument (integer) outside function:", immutable_num)

# Call the function
demonstrate_mutable_immutable()
```

```python
# Ans. 8

def demonstrate_logical_operators():
    # Define some boolean variables
    a = True
    b = False
    c = True

    # AND operator: returns True if both operands are True
    print("AND Operator:")
    print(f"a and b = {a and b}")  # False, because b is False
    print(f"a and c = {a and c}")  # True, because both a and c are True

    # OR operator: returns True if at least one operand is True
    print("\nOR Operator:")
    print(f"a or b = {a or b}")  # True, because a is True
    print(f"b or c = {b or c}")  # True, because c is True

    # NOT operator: inverts the truth value
    print("\nNOT Operator:")
    print(f"not a = {not a}")  # False, because a is True
    print(f"not b = {not b}")  # True, because b is False

# Call the function to demonstrate logical operators
demonstrate_logical_operators()
```

```python
# Ans. 9

def convert_input():
    # Get user input as a string
    user_input = input("Enter a value: ")

    # Convert the string to an integer
    try:
        int_value = int(user_input)
        print(f"Integer value: {int_value}")
    except ValueError:
        print("Could not convert to an integer.")

    # Convert the string to a float
    try:
        float_value = float(user_input)
        print(f"Float value: {float_value}")
    except ValueError:
        print("Could not convert to a float.")

    # Convert the string to a boolean
    # First, we check for a few common truthy/falsy values
    if user_input.lower() in ['true', '1', 'yes']:
        bool_value = True
    elif user_input.lower() in ['false', '0', 'no']:
        bool_value = False
    else:
        # Default is False if input is not recognized
        bool_value = False
    print(f"Boolean value: {bool_value}")

# Call the function to convert user input
convert_input()
```

```python
# Ans. 10

def demonstrate_type_casting():
    # Original list with different data types
    original_list = [1, 2.5, '3', True]

    # Convert all elements to integers
    int_list = [int(item) if isinstance(item, (float, str)) else item for item in original_list]

    # Convert all elements to floats
    float_list = [float(item) if isinstance(item, (int, str)) else item for item in original_list]

    # Convert all elements to strings
    str_list = [str(item) for item in original_list]

    # Display the results
    print("Original List:", original_list)
    print("List with elements cast to integers:", int_list)
    print("List with elements cast to floats:", float_list)
    print("List with elements cast to strings:", str_list)

# Call the function to demonstrate type casting
demonstrate_type_casting()
```

```python
# Ans. 14

def reverse_string(input_string):
    reversed_string = ""  # Initialize an empty string to store the reversed string
    index = len(input_string) - 1  # Start from the last index of the string

    # Use a while loop to iterate through the string in reverse
    while index >= 0:
        reversed_string += input_string[index]  # Add the character to the reversed string
        index -= 1  # Move to the previous character

    return reversed_string

# Example usage
input_string = input("Enter a string to reverse: ")
print("Reversed string:", reverse_string(input_string))
```

```python
#Ans. 15

def factorial(number):
    # Initialize the result to 1 (since factorial of 0 or 1 is 1)
    result = 1

    # Ensure the number is non-negative
    if number < 0:
        return "Factorial is not defined for negative numbers"

    # Calculate the factorial using a while loop
    while number > 1:
        result *= number
        number -= 1  # Decrease the number by 1 at each step

    return result

# Get input from the user
num = int(input("Enter a number to calculate its factorial: "))
print(f"The factorial is: {factorial(num)}")
```
